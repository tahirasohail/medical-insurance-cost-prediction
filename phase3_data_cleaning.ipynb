```python
import pandas as pd

df = pd.read_csv("data/insurance.csv")
print("Dataset loaded. Shape:", df.shape)
df.head()
```

```python
print("Missing values per column:")
print(df.isnull().sum())

print("\nTotal missing values in dataset:", df.isnull().sum().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())

duplicate_rows = df[df.duplicated()]
print("\nDuplicate rows found:")
duplicate_rows
```

```python
rows_before = df.shape[0]
df = df.drop_duplicates()
rows_after = df.shape[0]

print(f"Rows before removing duplicates: {rows_before}")
print(f"Rows after removing duplicates: {rows_after}")
print(f"Duplicates removed: {rows_before - rows_after}")
```

```python
print("Data types of each column:")
print(df.dtypes)
```

```python
categorical_cols = df.select_dtypes(include='object').columns.tolist()
print("Categorical columns:", categorical_cols)

for col in categorical_cols:
    print(f"\n{col} -> unique values: {df[col].unique()}")
```

```python
print("BEFORE preprocessing:")
print("Shape:", df.shape)
df.head()
```

```python
df['sex'] = df['sex'].map({'male': 0, 'female': 1})
df['smoker'] = df['smoker'].map({'no': 0, 'yes': 1})

print("sex and smoker encoded (0/1).")
df[['sex', 'smoker']].head()
```

```python
df = pd.get_dummies(df, columns=['region'], drop_first=True)

print("region one-hot encoded.")
df.head()
```

```python
print("AFTER preprocessing:")
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData types:")
print(df.dtypes)
df.head()
```

```python
print("Final missing values check:", df.isnull().sum().sum())
print("Final duplicate check:", df.duplicated().sum())
print("Final shape:", df.shape)
print("\nAll columns are now numeric:")
print(df.dtypes)
```

```python
df.to_csv("data/insurance_cleaned.csv", index=False)
print("Cleaned dataset saved as data/insurance_cleaned.csv")
```
